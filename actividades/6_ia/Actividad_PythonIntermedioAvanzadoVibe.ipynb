## 0. Repaso rápido: variables y tipos
Una **variable** es un nombre que guarda un valor. Python reconoce el tipo según el valor.


```python
nombre = "Ana"        # texto  (str)
edad = 25             # entero (int)
altura = 1.65         # decimal (float)
es_estudiante = True  # verdadero/falso (bool)

print(nombre, edad, altura, es_estudiante)
print("Tipo de 'edad':", type(edad))
```

## 1. Listas
Una **lista** es una colección **ordenada** que puede **cambiar**. Se escribe entre corchetes `[ ]`.
Piensa en una lista de compras: tiene orden y puedes tachar o agregar.


```python
frutas = ["manzana", "banana", "cereza"]
print(frutas)
print("La lista tiene", len(frutas), "elementos")
```

### Indexación: acceder por posición
**En Python se cuenta desde 0.** El índice `-1` es el último.


```python
frutas = ["manzana", "banana", "cereza"]
print(frutas[0])    # primero
print(frutas[1])    # segundo
print(frutas[-1])   # ultimo
```

### Slicing: tomar un pedazo. `lista[inicio:fin]` llega hasta **antes** de `fin`.


```python
numeros = [10, 20, 30, 40, 50]
print(numeros[1:3])   # [20, 30]
print(numeros[:2])    # los dos primeros
print(numeros[2:])    # del indice 2 en adelante
```

### Modificar y crecer


```python
frutas = ["manzana", "banana", "cereza"]
frutas[0] = "mango"        # cambiar
frutas.append("naranja")   # agregar al final
frutas.remove("banana")    # quitar por valor
print(frutas)
```

## 2. Tuplas
Como una lista, pero **no se puede modificar** (es *inmutable*). Se escribe entre paréntesis `( )`.
Sirve para datos que no deberían cambiar, como una coordenada `(latitud, longitud)`.


```python
punto = (4.6, -74.1)
print("Latitud:", punto[0])
print("Longitud:", punto[1])
```

```python
punto = (4.6, -74.1)
punto[0] = 5.0   # NO permitido en una tupla
```

## 3. Diccionarios
Guarda pares **clave : valor**. Accedes por **nombre** (la clave), no por posición. Se escribe entre llaves `{ }`.


```python
persona = {"nombre": "Ana", "edad": 25, "ciudad": "Quito"}
print(persona["nombre"])
print(persona["edad"])
```

## 4. Operaciones e indexación
### Operadores que usaras todo el tiempo


```python
a = 10
b = 3
print("Suma:", a + b)
print("Resta:", a - b)
print("Multiplicacion:", a * b)
print("Division:", a / b)
print("Residuo (modulo):", a % b)   # lo que sobra de dividir
print("¿a es mayor que b?", a > b)
```

## 5. Condicionales: `if` / `elif` / `else`
Sirven para **tomar decisiones**: ejecutar un bloque solo **si** se cumple una condición.
La condición es algo que da `True` o `False`. **La indentación (sangría) importa**: lo que está dentro del `if` va corrido a la derecha.


```python
edad = 20
if edad >= 18:
    print("Es mayor de edad")
else:
    print("Es menor de edad")
```

### Varias opciones con `elif`


```python
nota = 75
if nota >= 90:
    print("Excelente")
elif nota >= 60:
    print("Aprobado")
else:
    print("Reprobado")
```

### Combinar condiciones con `and`, `or`, `not`


```python
edad = 25
tiene_entrada = True

if edad >= 18 and tiene_entrada:
    print("Puede entrar")

if not tiene_entrada:
    print("Necesita comprar entrada")
```

## 6. Bucles `for`: repetir sin copiar y pegar
Un bucle `for` repite un bloque para cada elemento de una colección. Ya lo usamos; ahora a fondo.

### `range()` genera una secuencia de números


```python
for i in range(5):      # 0, 1, 2, 3, 4
    print(i)
```

```python
for i in range(1, 6):   # del 1 al 5
    print("Vuelta numero", i)
```

### Patrón "acumulador": ir sumando dentro del bucle


```python
suma = 0
for n in [10, 20, 30, 40]:
    suma = suma + n
print("La suma es", suma)
```

### `for` + `if`: recorrer y decidir


```python
numeros = [1, 2, 3, 4, 5, 6]
for n in numeros:
    if n % 2 == 0:
        print(n, "es par")
    else:
        print(n, "es impar")
```

## 9. Lectura e interpretación de errores
Equivocarse es **normal**. Python avisa con un **traceback**. La clave:
- **Lee la ÚLTIMA línea primero**: dice el **tipo de error** y una pista.
- Luego mira **en qué línea** ocurrió.

Cada celda **falla a propósito**: ejecútala, lee el mensaje y compáralo con la explicación.


### `NameError` — usaste un nombre que no existe


```python
print(mensaje)   # 'mensaje' nunca fue definido
```

`NameError: name 'mensaje' is not defined` → usaste una variable que no creaste (¿un typo? ¿no ejecutaste una celda anterior?).


### `TypeError` — mezclaste tipos incompatibles


```python
edad = 25
print("Tengo " + edad + " años")   # texto + numero no se puede
```

`TypeError` → intentaste unir texto con número. Solución: `str(edad)`.


### `IndexError` — pediste una posición que no existe


```python
frutas = ["mango", "cereza"]
print(frutas[5])   # solo hay posiciones 0 y 1
```

`IndexError: list index out of range` → el índice se pasa del tamaño de la lista.


### `KeyError` — pediste una clave que no existe


```python
persona = {"nombre": "Ana"}
print(persona["edad"])   # 'edad' no esta
```

`KeyError: 'edad'` → esa clave no está en el diccionario.
